```python
import torch
import random
import numpy as np
import pandas as pd
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW  # <--- FIXED IMPORT
from transformers import BartTokenizer, BartForConditionalGeneration, get_linear_schedule_with_warmup
from sentence_transformers import SentenceTransformer, util
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from tqdm.auto import tqdm

# ==========================================
# 1. DATA PREPARATION
# ==========================================

# ==========================================
# 2. CONFIGURATION & AUGMENTATION
# ==========================================

BATCH_SIZE = 16
EPOCHS = 15  # Increased epochs for better convergence
LEARNING_RATE = 2e-5
MAX_LEN = 32
AUGMENTATION_FACTOR = 20  # Generate 20 variants for every real sentence

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Helper to shuffle words (Synthetic noise generation)
def shuffle_sentence(sentence):
    words = sentence.split()
    random.shuffle(words)
    return " ".join(words)

# Create Training Data (Heavy Augmentation)
train_inputs = []
train_targets = []

print("Augmenting data...")
for sentence in merged_sentences:
    # Augment: 20 random shuffles per sentence
    for _ in range(AUGMENTATION_FACTOR):
        shuffled = shuffle_sentence(sentence)
        train_inputs.append(shuffled)
        train_targets.append(sentence)

# Create Validation Data
# We reserve the last 10% of merged_sentences for pure validation to test generalization
split_idx = int(len(merged_sentences) * 0.9)
val_gt = merged_sentences[split_idx:]
val_inputs = [shuffle_sentence(s) for s in val_gt] # Synthetic shuffle for val

# Filter training set to exclude validation sentences to prevent leakage
train_clean_inputs = []
train_clean_targets = []
val_set_hashes = set(val_gt)

for inp, tgt in zip(train_inputs, train_targets):
    if tgt not in val_set_hashes:
        train_clean_inputs.append(inp)
        train_clean_targets.append(tgt)

print(f"Training Samples (Augmented): {len(train_clean_inputs)}")
print(f"Validation Samples: {len(val_inputs)}")

# ==========================================
# 3. DATASET CLASS
# ==========================================

class ReconstructionDataset(Dataset):
    def __init__(self, inputs, targets, tokenizer, max_len=32):
        self.inputs = inputs
        self.targets = targets
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, idx):
        input_text = self.inputs[idx]
        target_text = self.targets[idx]

        # Tokenize Input
        input_enc = self.tokenizer(
            input_text, max_length=self.max_len, padding='max_length',
            truncation=True, return_tensors='pt'
        )

        # Tokenize Target
        target_enc = self.tokenizer(
            target_text, max_length=self.max_len, padding='max_length',
            truncation=True, return_tensors='pt'
        )

        return {
            'input_ids': input_enc['input_ids'].flatten(),
            'attention_mask': input_enc['attention_mask'].flatten(),
            'labels': target_enc['input_ids'].flatten(),
            'target_text': target_text,
            'input_text': input_text
        }

# ==========================================
# 4. MODEL SETUP (BART)
# ==========================================

model_name = "facebook/bart-base"
tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name).to(device)

train_dataset = ReconstructionDataset(train_clean_inputs, train_clean_targets, tokenizer)
val_dataset = ReconstructionDataset(val_inputs, val_gt, tokenizer)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE)

optimizer = AdamW(model.parameters(), lr=LEARNING_RATE)
scheduler = get_linear_schedule_with_warmup(
    optimizer, num_warmup_steps=100, num_training_steps=len(train_loader) * EPOCHS
)

# Semantic Similarity Model (Fast version)
sim_model = SentenceTransformer('all-MiniLM-L6-v2')

# ==========================================
# 5. TRAINING LOOP WITH METRICS
# ==========================================

# Results storage
history = []

print("\nStarting Training...")
print("-" * 90)
print(f"{'Epoch':<6} | {'Train Loss':<12} | {'Val Loss':<10} | {'BLEU':<8} | {'Cos Sim':<8}")
print("-" * 90)

for epoch in range(EPOCHS):
    # --- Training ---
    model.train()
    total_train_loss = 0
    for batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}", leave=False):
        optimizer.zero_grad()
        
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels = batch['labels'].to(device)

        outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        total_train_loss += loss.item()

        loss.backward()
        optimizer.step()
        scheduler.step()

    avg_train_loss = total_train_loss / len(train_loader)

    # --- Validation & Metrics ---
    model.eval()
    total_val_loss = 0
    predictions = []
    references = []
    
    with torch.no_grad():
        for batch in val_loader:
            input_ids = batch['input_ids'].to(device)
            attention_mask = batch['attention_mask'].to(device)
            labels = batch['labels'].to(device)

            # Calculate Loss
            outputs = model(input_ids=input_ids, attention_mask=attention_mask, labels=labels)
            total_val_loss += loss.item()

            # Generate Text for Metrics
            generated_ids = model.generate(input_ids=input_ids, attention_mask=attention_mask, max_length=MAX_LEN)
            decoded_preds = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)
            
            predictions.extend(decoded_preds)
            references.extend(batch['target_text'])

    avg_val_loss = total_val_loss / len(val_loader)

    # Calculate BLEU (Corpus Level for speed & accuracy)
    bleu_scores = []
    smooth = SmoothingFunction().method1
    for pred, ref in zip(predictions, references):
        bleu_scores.append(sentence_bleu([ref.split()], pred.split(), smoothing_function=smooth))
    avg_bleu = np.mean(bleu_scores)

    # Calculate Cosine Similarity
    # Encode all preds and refs at once
    pred_embeddings = sim_model.encode(predictions, convert_to_tensor=True, show_progress_bar=False)
    ref_embeddings = sim_model.encode(references, convert_to_tensor=True, show_progress_bar=False)
    cosine_scores = util.cos_sim(pred_embeddings, ref_embeddings)
    # Extract diagonal (pairwise match)
    pairwise_cosine = torch.diag(cosine_scores).mean().item()

    # Log
    history.append({
        'Epoch': epoch + 1,
        'Train Loss': avg_train_loss,
        'Val Loss': avg_val_loss,
        'Bleu': avg_bleu,
        'Cos Sim': pairwise_cosine
    })

    print(f"{epoch+1:<6} | {avg_train_loss:<12.4f} | {avg_val_loss:<10.4f} | {avg_bleu:<8.4f} | {pairwise_cosine:<8.4f}")

# ==========================================
# 6. SHOW INSTANCES
# ==========================================

print("\n" + "="*80)
print("TEST SAMPLES (Scattered vs Ground Truth vs Reconstructed)")
print("="*80)

# Pick 10 random samples from validation
sample_indices = random.sample(range(len(references)), min(10, len(references)))

for idx in sample_indices:
    print(f"Input (Scattered): {val_inputs[idx]}")
    print(f"Ground Truth     : {val_gt[idx]}")
    print(f"Model Output     : {predictions[idx]}")
    print("-" * 50)

# ==========================================
# 7. FINAL TABLE
# ==========================================
print("\nFinal Metrics Summary:")
df_results = pd.DataFrame(history)
print(df_results)
```
